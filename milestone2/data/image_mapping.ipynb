#### Image URL Mapping

Match `product_title` from processed.csv (A) with image_url.csv (B) to add product images.

```python
import pandas as pd

csv_a = pd.read_csv('../../milestone1/output/processed.csv')
csv_b = pd.read_csv('./image_url.csv')

print(f'CSV A (processed): {len(csv_a)} rows, {csv_a["product_title"].nunique()} unique products')
print(f'CSV B (image_url): {len(csv_b)} rows, {csv_b["product_title"].nunique()} unique products')
```

```python
image_lookup = csv_b.drop_duplicates(subset='product_title').set_index('product_title')['image_url']

csv_a['image'] = csv_a['product_title'].map(image_lookup).fillna('Not included')

matched = (csv_a['image'] != 'Not included').sum()
total = len(csv_a)
unique_matched = csv_a[csv_a['image'] != 'Not included']['product_title'].nunique()
unique_total = csv_a['product_title'].nunique()

print(f'Matched: {matched:,}/{total:,} rows ({matched/total:.1%})')
print(f'Unique products matched: {unique_matched}/{unique_total}')
print(f'Not included: {total - matched:,} rows')
```

```python
print('Sample matched:')
display(csv_a[csv_a['image'] != 'Not included'][['product_title', 'brand_name', 'image']].drop_duplicates().head())

print('\nSample not matched:')
not_matched = csv_a[csv_a['image'] == 'Not included']['product_title'].unique()
print(f'{len(not_matched)} unique products not matched:')
for t in not_matched[:10]:
    print(f'  - {t}')
```

```python
csv_a.to_csv('processed_with_images.csv', index=False)
print(f'Saved processed_with_images.csv ({len(csv_a):,} rows)')
print(f'Columns: {list(csv_a.columns)}')
```
